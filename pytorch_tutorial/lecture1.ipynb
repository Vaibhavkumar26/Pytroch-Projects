The rule for matrix multiplication for more that 3-dimensional matrix 
    
    for 3D+ tensors, matrix multiplication (torch.matmul or @) splits the shape into two parts: the last two dimensions (the actual matrix) and everything before that (the batch dimensions)

    Rule 1 : Last two dims → standard matrix-multiply rule still applies: c1 == r2 (non-negotiable, same as 2D) (..... , r1 , c1)  (...... , r2 , c2)

        (....... , r1 , c2)

    Rule 2 : Batch dims (everything before the last two) → must be broadcastable with each other using the same broadcasting rules from before (equal, or one of them is 1)

```python
## Reshaping , stacking ,  viewing and stacking Tensors
# Reshaping - reshapes an input tensor to a defined shape

# View - Return a view of an input tensor of certain shape but keep the same memory
# Stacking - Combine multiple tensors onn top of each other (vs stack ) or side by side (hstack)


# Reshaping 
torch.manual_seed(42)

a = torch.arange(1 , 10)
print(f" the original A value {a}")
a=a.reshape( 3 , 3) # 9 elements have being rearranged
print(f"the element rearranged after some times {a}")
a.reshape( 3 , -1) ##Imp feature

## flatten images for a Linear 

image = torch.rand(size=(3 , 2 , 2 , 2))
print(f"The image before we flatten the dimensions {image}")

image=image.reshape(image.size(0) , -1)
print(f"After the change flatten as occured the shape {image.shape}")
print(image)

```

```python
image_2 = torch.rand(size=(3 , 1 , 2 , 2))
print(f"the image_2 {image_2}")

image_2 =image_2.reshape(image_2.size(0) , image_2.size(1) , -1)
print(f"After the flatten as occured {image_2.shape}")
print(image_2)


image_3 = torch.rand(size=(5,))
print(f"After the 3rd image is found {image_3}")
image_3 = image_3.reshape(-1 , 1)

print(f"After the new dimension added shape : -{image_3.shape}")
print(image_3)

#reshape() (like view(), transpose(), stack(), etc.) does not modify the tensor in place — it returns a brand-new
#  tensor/view. If you don't capture that return value with =, the operation still runs, but the result just... evaporates.
```

```python
# Stacking - torch.stack(tensors, dim=0)

# Combines tensors along a brand-new dimension. All tensors must have identical shapes.

a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])
c = torch.tensor([7, 8, 9])

horizontal_stack = torch.stack([a, b, c], dim=0)  # (3, 3): each tensor becomes a row
vertical_stack = torch.stack([a, b, c], dim=1)  # (3, 3): each tensor becomes a column

print(f"The horizontal stack:\n{horizontal_stack}")
print(f"The vertical stack:\n{vertical_stack}")
```
